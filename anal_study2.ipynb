```python
df = pd.read_csv('study-2/waist_impressions_DE_random_50_users_v3.csv', low_memory=False)
df
```

```python
# convert csv to json for a bette visualization
import pandas as pd
import json
from json import JSONDecodeError

def try_parse_json(value):
    if pd.isna(value) or not isinstance(value, str):
        return value
    try:
        return json.loads(value)
    except JSONDecodeError:
        return value

# Apply JSON parsing to all columns
for col in df.columns:
    df[col] = df[col].apply(try_parse_json)

# Convert to JSON
df.to_json('output.json', orient='records', indent=4)
```

```python
df_head = df[:1000]
df_head.to_json('output_head.json', orient='records', indent=4)
```
