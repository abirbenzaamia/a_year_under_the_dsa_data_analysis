```python
from collections import defaultdict


def extract_waist_targeting_data(data):
    result = {}
    targeting_data_list = defaultdict(list)
    #targeting_data_list = []
    if isinstance(data, dict):  
        #print(advertiser_name)  # For debugging
        waist = data['data']
        waist_targeting_data = waist['waist_targeting_data']
        #print(f"Extracting targeting data from waist_targeting_data: {waist_targeting_data}")
        for item in waist_targeting_data:
            key = item['waist_ui_type']            
            value = item['serialized_data']
            #print(f"Extracted key: {key}, value: {value}")
            try:
                targeting_data_list[key].append(value)
            except KeyError:
                print(f"KeyError for key: {key}, value: {value}")            
        #print(f"Extracted targeting data: {targeting_data_list}")

        for key, value in targeting_data_list.items():
            result[key] = value

        return result
    else: 
        print(f"Unknown data type: {type(data)}")
        return None
    

df_facebook['targeting_data'] = df_facebook['provider_data.waist'].apply(extract_waist_targeting_data)
```

```python
# convert english str to list

import re
import json 
import ast
def force_convert_to_list(str_data):
    # str_data = str_data.strip()
    # if str_data.startswith('[') and str_data.endswith(']'):
    #     str_data = str_data[1:-1]
    #print(f"Converting to list: {str_data}")
    try:
       return ast.literal_eval(str_data)  
    except:
        protected = re.sub(r"(\w)'([a-z])", r"\1’", str_data)  # Replace with curly apostrophe
    
        # Step 2: Convert remaining single quotes to double quotes
        standardized = protected.replace("'", '"')

        # Step 3: Restore protected apostrophes
        final_str = standardized.replace("’", "'")

        try:
            return ast.literal_eval(final_str)
        except (SyntaxError, ValueError):
            try:
                # Fallback for malformed strings
                return ast.literal_eval(final_str.replace('"', "'"))
            except:
                return []  # Ultimate fallback


df_insta['english_waist_converted'] = df_insta['english_waist'].apply(
    lambda x: force_convert_to_list(x) if isinstance(x, str) else x
)
```

```python
# extract age from the ad library
import re

def extract_age_from_waist(value):

    age_pattern = re.compile(
    r'(?:'
    r'(?:aged|ages|age range from) \d+[-+]?\d* to \d+[-+]?\d*'
    r'|(?:an age of|aged|age of|age|age as)? \d+[-+]?\d*(?: years)? (?:or|and) (?:older|over|above)'
    r'|(?:age|aged) between \d+[-+]?\d* and \d+[-+]?\d*'
    r'|people (?:aged|over the age of) \d+'
    r'|set their age to \d+'
    r'|from the age of \d+'
    r'|are between the ages of \d+[-+]?\d* and \d+[-+]?\d*'
    r'|over \d+[-+]?\d* years old'
    r')',
    re.IGNORECASE
     )
   
    value_lower = value.lower()
    if re.search(age_pattern, value_lower):
        #print(f"Matched age pattern: {value}")
        return value
   


def process_age(row):
    data = row['english_waist_converted']
    age = None
    if isinstance(data, list):
        for item in data:
            age = extract_age_from_waist(item)
            if age:
                break
            
        return age
    else:     
        return None

df_insta['waist_age'] = df_insta.apply(process_age, axis=1)


```
