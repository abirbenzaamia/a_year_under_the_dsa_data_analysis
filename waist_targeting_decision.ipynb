```python
## calculate the frequency of each type with the most common value

from collections import defaultdict, Counter

df_instagram_explanations_freq = pd.DataFrame(columns=['explanation', 'freq'])
# Initialize dictionaries
category_count = defaultdict(int)
explanation_by_category = defaultdict(list)
targeting_info_count = 0
ad_explanations_nb = 0
ad_explanations_nb_count =  defaultdict(int)
granularity_count_per_category = defaultdict(lambda: defaultdict(int))
# Count frequency and group explanations
ad_explanations_by_granualrity =  defaultdict(int)


for index, row in df_instagram.iterrows():    
    data = row['targeting_categories_category_decision']
    if isinstance(data,dict):
        ad_explanations_nb +=1 
        for item in data:
                category_count[item] += len(data[item])
                for granularity in data[item]:
                    granularity_count_per_category[item][granularity] +=1
                    targeting_info_count +=1
                    ad_explanations_by_granualrity[granularity] +=1
                

print("Targeting information count:", targeting_info_count)

print("\nGranularity Count per Category:")
for category, counts in granularity_count_per_category.items():
    print(f"Category: {category} --------- ")
    for granularity, freq in counts.items():
        #print(f"  Granularity Level {granularity}: used {freq} times")
        print(f"  Granularity Level {granularity}: used {freq/category_count[category] * 100:.2f} %")



for granularity, freq in ad_explanations_by_granualrity.items():
     print(f"Overall Granularity Level {granularity}: used {freq/targeting_info_count * 100:.2f} %")
   
```
