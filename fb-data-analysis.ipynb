```python
import ast
import json
import pandas as pd
import numpy as np

def safe_str_to_dict(text):
    """Convert string dict to Python dict with comprehensive error handling"""
    if pd.isna(text) or not isinstance(text, str):
        return {}
    
    # First try json.loads (handles proper JSON)
    try:
        return json.loads(text)
    except json.JSONDecodeError:
        pass
    
    # Then try ast.literal_eval (handles Python literals)
    try:
        return ast.literal_eval(text)
    except (ValueError, SyntaxError):
        pass
    
    # Fallback for malformed strings
    try:
        # Handle common issues in string representations
        text = (text.replace("'", '"')
                .replace('None', 'null')
                .replace('True', 'true')
                .replace('False', 'false'))
        return json.loads(text)
    except:
        return text  # Return empty dict if all attempts fail

```

```python
# extract pattern is waist 


import pandas as pd
import ast
from collections import defaultdict

def extract_targeting_patterns(targeting_dict):
    """
    Extract all targeting patterns from nested dictionary structure
    Returns a flattened dictionary with all targeting information
    """
    result = defaultdict(list)
    
    # Extract INTERESTS data
    if 'INTERESTS' in targeting_dict:
        interests = targeting_dict['INTERESTS']
        if isinstance(interests, dict) and 'interests' in interests:
            for interest in interests['interests']:
                if isinstance(interest, dict) and 'name' in interest:
                    result['interests'].append(interest['name'])
    
    # Extract AGE_GENDER data
    if 'AGE_GENDER' in targeting_dict:
        age_gender = targeting_dict['AGE_GENDER']
        if isinstance(age_gender, dict):
            result['age_min'] = age_gender.get('age_min')
            result['age_max'] = age_gender.get('age_max')
            result['gender'] = age_gender.get('gender')
    
    # Extract LOCATION data
    if 'LOCATION' in targeting_dict:
        location = targeting_dict['LOCATION']
        if isinstance(location, dict):
            result['location_name'] = location.get('location_name')
            result['location_granularity'] = location.get('location_granularity')
            result['location_code'] = location.get('location_code')
    
    # Extract other known targeting types
    for key in ['CUSTOM_AUDIENCES', 'RELATIONSHIP_STATUS', 'WORK_EMPLOYERS']:
        if key in targeting_dict:
            result[key.lower()] = targeting_dict[key]
    
    # Convert defaultdict to regular dict
    return dict(result)
```

```python
import pandas as pd
import ast
from collections import defaultdict

def get_unique_dict_keys(df, dict_columns):
    """
    Extract all unique keys from dictionary columns in a DataFrame
    
    Args:
        df: pandas DataFrame
        dict_columns: list of column names containing dictionaries
        
    Returns:
        Dictionary mapping column names to sets of unique keys
    """
    unique_keys = defaultdict(set)
    
    for col in dict_columns:
        # Ensure column exists
        if col not in df.columns:
            continue
            
        # Process each row
        for val in df[col]:
            if pd.isna(val):
                continue
                
            # Convert string to dict if needed
            if isinstance(val, str):
                try:
                    val = ast.literal_eval(val)
                except:
                    continue
                    
            # Collect keys if it's a dictionary
            if isinstance(val, dict):
                unique_keys[col].update(val.keys())
    
    return {col: sorted(keys) for col, keys in unique_keys.items()}

# Usage example:
unique_keys = get_unique_dict_keys(df_facebook, ['provider_data.waist'])
for col, keys in unique_keys.items():
    print(f"Column '{col}' has keys: {keys}")
```

```python
# we have 1500 row data where we do have only targeting_categoriesand do not have provider_data.waist 


def get_nested_keys(d, prefix=''):
    """Recursively get all nested dictionary keys with dot notation"""
    keys = []
    if isinstance(d, dict):
        for k, v in d.items():
            new_prefix = f"{prefix}.{k}" if prefix else k
            keys.append(new_prefix)
            if isinstance(v, dict):
                keys.extend(get_nested_keys(v, new_prefix))
    return keys

def get_all_nested_keys(df, dict_columns):
    """Get all unique nested keys from dictionary columns"""
    all_keys = defaultdict(set)
    
    for col in dict_columns:
        if col not in df.columns:
            continue
            
        for val in df[col]:
            if pd.isna(val):
                continue
                
            if isinstance(val, str):
                try:
                    val = ast.literal_eval(val)
                except:
                    continue
                    
            if isinstance(val, dict):
                for key in get_nested_keys(val):
                    all_keys[col].add(key)
    
    return {col: sorted(keys) for col, keys in all_keys.items()}

# Usage:
nested_keys = get_all_nested_keys(df_facebook, ['provider_data.waist'])
for col, keys in nested_keys.items():
    print(f"Column '{col}' nested keys: {keys}")
```

```python
import pandas as pd
import ast
from collections import defaultdict

def get_first_occurrence_indexes(df, dict_columns):
    """
    Get the first index where each dictionary key appears
    
    Args:
        df: pandas DataFrame
        dict_columns: list of column names containing dictionaries
        
    Returns:
        Dictionary {column_name: {key: first_index}}
    """
    first_occurrences = defaultdict(dict)
    seen_keys = defaultdict(set)
    
    for col in dict_columns:
        if col not in df.columns:
            continue
            
        for idx, val in df[col].items():
            if pd.isna(val):
                continue
                
            # Convert string to dict if needed
            if isinstance(val, str):
                try:
                    val = ast.literal_eval(val)
                except:
                    continue
                    
            if isinstance(val, dict):
                for key in val.keys():
                    if key not in seen_keys[col]:
                        seen_keys[col].add(key)
                        first_occurrences[col][key] = idx
    
    return dict(first_occurrences)

# Usage:
first_occurrences = get_first_occurrence_indexes(df_facebook, ['provider_data.waist'])
for col, keys in first_occurrences.items():
    print(f"\nColumn: {col}")
    for key, idx in keys.items():
        print(f"Key '{key}' first appears at index: {idx}")
        print("Example value:", df_facebook.at[idx, col])
        print("date",df_facebook.at[idx, 'month'],df_facebook.at[idx, 'year'])
        print("-----------")
```

```python
def get_first_nested_occurrences(df, dict_columns):
    """Get first index for nested keys (dot notation)"""
    first_occurrences = defaultdict(dict)
    seen_keys = defaultdict(set)
    
    def collect_keys(d, prefix=''):
        keys = []
        if isinstance(d, dict):
            for k, v in d.items():
                full_key = f"{prefix}.{k}" if prefix else k
                keys.append(full_key)
                if isinstance(v, dict):
                    keys.extend(collect_keys(v, full_key))
        return keys
    
    for col in dict_columns:
        if col not in df.columns:
            continue
            
        for idx, val in df[col].items():
            if pd.isna(val):
                continue
                
            if isinstance(val, str):
                try:
                    val = ast.literal_eval(val)
                except:
                    continue
                    
            if isinstance(val, dict):
                for key in collect_keys(val):
                    if key not in seen_keys[col]:
                        seen_keys[col].add(key)
                        first_occurrences[col][key] = idx
    
    return dict(first_occurrences)

# Usage:
nested_occurrences = get_first_nested_occurrences(df_facebook, ['provider_data.waist'])
for col, keys in nested_occurrences.items():
    print(f"\nColumn: {col}")
    for key, idx in keys.items():
        print(f"Nested key '{key}' first appears at index: {idx}")
        print("date",df_facebook.at[idx, 'month'],df_facebook.at[idx, 'year'])
        print("-----")
```

```python
# now we split the data into 3 sub df where in each we have same patterns of ad explanations 

def split_timeline_facebook(df, col, target_key):
   
    if col not in df.columns:
        raise ValueError(f"Column '{col}' not found in DataFrame")
    
    # Convert string to dict if needed
    def safe_parse(x):
        if pd.isna(x):
            return {}
        if isinstance(x, str):
            try:
                return ast.literal_eval(x) if x.strip() else {}
            except:
                return {}
        return x if isinstance(x, dict) else {}
    
    # Create mask for rows containing the key
    has_key_mask = df[col].apply(
        lambda x: target_key in safe_parse(x)
    )
    
    return df[has_key_mask].copy(), df[~has_key_mask].copy()
```

```python
def split_by_nested_key(df, dict_column, target_path):
    """
    Split by nested key using dot notation (e.g., 'INTERESTS.interests')
    """
    def has_nested_key(d, path):
        keys = path.split('.')
        for key in keys:
            if not isinstance(d, dict) or key not in d:
                return False
            d = d[key]
        return True
    
    def safe_parse(x):
        if pd.isna(x):
            return {}
        if isinstance(x, str):
            try:
                return ast.literal_eval(x) if x.strip() else {}
            except:
                return {}
        return x if isinstance(x, dict) else {}
    
    has_key_mask = df[dict_column].apply(
        lambda x: has_nested_key(safe_parse(x), target_path)
    )
    
    return df[has_key_mask].copy(), df[~has_key_mask].copy()
```
