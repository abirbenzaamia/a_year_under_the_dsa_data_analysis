```python
#classify english_waist 

# applying my own clustering ad trasparency information

import re

def classify_waist_youtube(value,advertiser):

    targeting_patterns = {
    
    # BEHAVIORAL BASED
    'behavioral' : [
        r'(?:based on websites|due to websites) you(?: have|\'ve|\'e)? visited',
        r'google activities and visits to other websites',
        r'your (?:activity|activities)(?:,)? (?:where|while) (?:you were signed|signed|you were logged|logged) (?:in to|into) (?:google|your google account)',
    ], 

    # INTEREST BASED
    'interest' : [
        r'your interests as assessed by google',
        r'google\'s estimation of your interests',
        r'the interests that google estimates you have',
        r'your interests according to',
        r'based on (google\' assessment of your interests|the assessment of your interests by google)',
        r'your interests as estimated by google',
        r'the interests that google assesses you(?: to)? have',
        r'the interests you have according to google\'s assessment',
        r'google\'s estimation of your areas of interest',
        r'the interests(?: that)? google (thinks|believes|assumes|estimates|assesses) you(?: to)? have',
        r'the interests you have according to google',
        r'the interests that google associates with you',

    ],

    # DEMOGRAPHIC TARGETING
    #Details like age, gender, language, parental status, and household income. These are based on info users have shared with Google or that's been guessed.
    'demographic':  [
         r'your gender|your age|home ownership status|parental status|education degree'
        ],

    # HASHED LIST 
    'costum audience': [
       r'information you(?: have)? (provided|gave) to the advertiser',
       r'the advertiser\'(?:s)? goal to reach new customers who have not yet purchased from them',
        r'the advertiser\'s goal to reach new customers who haven\'t bought from them before',
        r'the advertiser\’s interest in reaching new customers who haven\’t bought something from them before',
        r'the advertiser\'s goal of reaching new customers who have not yet purchased from them',
        r'the goal of the advertiser to reach new customers who have not bought anything from them yet',
        r'the goal of reaching new customers who have not yet purchased from them',
        r'the advertiser\' goal to reach new customers',
        r'the advertiser\' goal of reaching new customers',
        r'the advertiser\' goal is to reach new customers',

    ],
    # RETARGETING
    'retargeting': [
        r'your use of the advertiser\' website or app',
        r'your activity on youtube related to the advertiser'
        ],

    # LOOKALIKE TARGETING
    'lookalike': [
        r'your similarity to(?: groups of)? people(?: that| whom)? the advertiser is (?:trying|attempting) to reach',
        r'your similarity to demographic groups that the advertiser is trying to reach',
        

        ],
    # CONTEXTUAL TARGETING 
    'contextual': [
           r'the video you(?: are|\'re) watching',
           r'your current search terms',
        ],
    
    #GEOLOCATION TARGETING
    'location': [
        r'general location', 
        r'your approximate current location',
        r'your approximate current location (?:as estimated by google|according to Google\' estimation)',
        r'your general (geographic|geographical) location',
    ],

    # TIME
    'time': [
        r'time of day', 
        r'the hour of the day',
        r'schedule'
    ],
    }
    
    ##print(type(value))
    value_lower = value.lower()
    for type_name, regex_list in targeting_patterns.items():
        for pattern in regex_list:
            if re.search(pattern, value_lower):
                return type_name
            
    #print(advertiser)
    print(f"Unknown pattern: {value}")
    print(advertiser)
  
    return None


def process_targeting_categories(row):
    data = row['english_waist_corrected']
    advertiser_name = row['provider_data.rawlog_id']

    if isinstance(data, list):
        #print(advertiser_name)  # For debugging
        return [{classify_waist_youtube(item,advertiser_name): item} for item in data]
    elif isinstance(data, str):
        return {classify_waist_youtube(data,advertiser_name): data}
    else:
        return None

df_youtube['targeting_categories_english'] = df_youtube.apply(process_targeting_categories, axis=1)
```
