```python
# now we do matching of extracted taregting data values with user-facing ad expalantions 

def set_user_facing_waist_2024(row):
    user_facing_waist = list()
    targeting_data = row['extracted_taregting_data_values']
    for item in targeting_data:

        if item == 'AGE_GENDER':
            # add gender 
            if targeting_data[item]['age_max'] == 53: 
                age_gender = f"set their age to {targeting_data[item]['age_min']+12} and older"
            else:
                age_gender = f"set their age between {targeting_data[item]['age_min']+12} and {targeting_data[item]['age_max']+12}"
            user_facing_waist.append(age_gender)

        elif item == 'LOCATION':
            if targeting_data[item]['location_geo_type'] == 'home':
                location = f"a primary location in {targeting_data[item]['location_name']}"
            else:
                location = f"People located in {targeting_data[item]['location_name']}"
            user_facing_waist.append(location)
            
        elif item == 'LOCALE':
            language = f"communicated in {', '.join(targeting_data[item]['locales'])}"
            user_facing_waist.append(language)
    
        elif item == 'INTERESTS':
            interests = f"people interested in {', '.join(targeting_data[item]['interests'])}"
            user_facing_waist.append(interests)

        elif item == 'CUSTOM_AUDIENCES_DATAFILE':
            if targeting_data[item]['match_keys'] == []:
                custom_audience = f"{targeting_data[item]['owner']} has uploaded a hashed list with you inforamtion on Facebook. This information was used to match with your data."
            else:
                match_key = targeting_data[item]['match_keys']
                custom_audience = f"{targeting_data[item]['owner']} has uploaded a hashed list with you inforamtion on Facebook. Your {', '.join(match_key)} was used to match with your data."
            user_facing_waist.append(custom_audience)

        elif item == 'CUSTOM_AUDIENCES_WEBSITE':
            custom_audience = f"People who have visited the website of {targeting_data[item]['owner']} or a partner"            
            user_facing_waist.append(custom_audience)

        elif item == 'CUSTOM_AUDIENCES_LOOKALIKE':
            custom_audience = f"People who resemble existing customers"
            user_facing_waist.append(custom_audience)

        elif item == 'CUSTOM_AUDIENCES_ENGAGEMENT_PAGE':
            custom_audience = f"People who have interacted with the Facebook account of {targeting_data[item]['owner']}"
            user_facing_waist.append(custom_audience)

        elif item == 'CUSTOM_AUDIENCES_ENGAGEMENT_IG':
            custom_audience = f"People who have interacted with the Instagram account of {targeting_data[item]['owner']}"
            user_facing_waist.append(custom_audience)

        elif item == 'DYNAMIC_RULE':
            continue

        elif item == 'CUSTOM_AUDIENCES_ENGAGEMENT_VIDEO':
            custom_audience = f"People who watched a video from {targeting_data[item]['owner']}'s Facebook page"
            user_facing_waist.append(custom_audience)

        elif item == 'BCT':
            bct = f"People who have been added to the {targeting_data[item]['bct_name']} category"
            user_facing_waist.append(bct)

        elif item == 'LOCAL_REACH':
            continue 

        elif item == 'ED_STATUS':
            education = f"set their education to {targeting_data[item]['edu_status']}"
            user_facing_waist.append(education)

        elif item == 'RELATIONSHIP_STATUS':
            relationship = f"People who have indicated {targeting_data[item]['relationship_status']} as their relationship status on Facebook"
            user_facing_waist.append(relationship)
        else:
            continue

    return list(user_facing_waist)
        
        
df_2024['user_facing_waist'] = df_2024.apply(set_user_facing_waist_2024, axis=1)

```

```python
# recalculate taregting_categories 

import re
from collections import defaultdict


def classify_waist_instagram(value):
    targeting_patterns = {
    
        'behavioral and interest' : [
          r'people interested in',
          r'people who have been added to the',
        ],

    # DEMOGRAPHIC TARGETING
    'demographic':  [
          r'set their age to',
          r'set their age between',
          r'communicated in',
          r'set their education to',
          r'as their relationship status',

        ],
    #HASHED LIST 
    'custom audience': [
         r'has uploaded a hashed list with you inforamtion',

    ],
    # RETARGETING
    'remarketing': [
         r'people who have visited the website of',
         r'people who have interacted with the facebook account',
         r'people who have interacted with the instagram account',
         r'people who watched a video from',
        ],

    # LOOKALIKE TARGETING
    'lookalike': [
          r'people who resemble existing customers',
        ],
    # CONTEXTUAL TARGETING  # No contextual in Insta
    
    # GEOLOCATION TARGETING 
    'location': [
          r'people located in',
          r'a primary location in',
    ]
    }

    
    ##print(type(value))
    value_lower = value.lower()
    for type_name, regex_list in targeting_patterns.items():
        for pattern in regex_list:
            if re.search(pattern, value_lower):
                return type_name
            
    
    with open("pattern_insta.txt", 'a', encoding='utf-8') as f:
        f.write(value_lower + "\n") 
    print(f"Unknown pattern: {value}")
    return None



def process_targeting_categories(row):
    classified_items = defaultdict(list)
    result = {}
    data = row['user_facing_waist']
    if isinstance(data, list):
        for item in data:
            targeting_type = classify_waist_instagram(item)
            classified_items[targeting_type].append(item)
            #print(classified_items)
        
        for classification, items in classified_items.items():
            result[classification] = items
                
        return result
    elif isinstance(data, str):
        return {classify_waist_instagram(data): data}
    else:     
        return None

df_fb_comb['targeting_categories_english'] = df_fb_comb.apply(process_targeting_categories, axis=1)
```
